```python
import pandas as pd


class NaiveBayes:

    def __init__(self):
        self.features = []
        self.classes = []
        self.prior = {}
        self.conditional = {}


    def train(self, data):


        target = data.columns[-1]

        self.features = list(data.columns[:-1])

        self.classes = data[target].unique().tolist()



        for cls in self.classes:

            class_count = len(data[data[target] == cls])

            self.prior[cls] = class_count / len(data)



        for feature in self.features:

            self.conditional[feature] = {}


            values = data[feature].unique()

            for value in values:

                self.conditional[feature][value] = {}

                for cls in self.classes:


                    class_data = data[data[target] == cls]


                    count = len(
                        class_data[class_data[feature] == value]
                    )


                    probability = count / len(class_data)

                    self.conditional[feature][value][cls] = probability

    def predict(self, test):

        probabilities = {}


        for cls in self.classes:


            probability = self.prior[cls]


            for feature in self.features:

                value = test[feature]

                probability *= (
                    self.conditional[feature][value][cls]
                )

            probabilities[cls] = probability

        prediction = max(
            probabilities,
            key=probabilities.get
        )

        return prediction, probabilities


    def display(self):

        print("\nFEATURES:")
        for feature in self.features:
            print(feature)

        print("\nPOSSIBLE VALUES:")

        for feature in self.features:
            print(feature, ":", list(self.conditional[feature].keys()))

        print("\nCLASSES:")
        print(self.classes)

        print("\nPRIOR PROBABILITIES:")

        for cls in self.classes:
            print(
                "P(" + cls + ") =",
                self.prior[cls]
            )

        print("\nCONDITIONAL PROBABILITIES:")

        for feature in self.features:

            print("\n", feature)

            for value in self.conditional[feature]:

                for cls in self.classes:

                    probability = self.conditional[
                        feature
                    ][value][cls]

                    print(
                        "P(" + feature + "=" + value +
                        " | " + cls + ") =",
                        probability
                    )


data = pd.DataFrame({

    "Attendance": [
        "Low",
        "Low",
        "Medium",
        "High",
        "High",
        "Medium",
        "Low",
        "Low"
    ],

    "Internal Marks": [
        "Low",
        "High",
        "Low",
        "Low",
        "High",
        "Low",
        "High",
        "Low"
    ],

    "Risk": [
        "Yes",
        "Yes",
        "Yes",
        "Yes",
        "No",
        "No",
        "No",
        "No"
    ]
})




model = NaiveBayes()

model.train(data)

model.display()


# =====================================================
# TEST INSTANCE
# =====================================================

test = {
    "Attendance": "Medium",
    "Internal Marks": "High"
}




prediction, probabilities = model.predict(test)


print("\nTEST INSTANCE:")
print(test)


print("\nCLASS PROBABILITIES:")

for cls in probabilities:

    print(
        "P(" + cls + " | Test) =",
        probabilities[cls]
    )


print("\nPREDICTED CLASS:", prediction)
```
