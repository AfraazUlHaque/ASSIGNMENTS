```python
print(df.isnull().sum())
print("Duplicates:", df.duplicated().sum())
df = df.drop_duplicates()
```

```python
X = df.iloc[:, :-1]
y = df.iloc[:, -1]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.10, random_state=42
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)
y_pred = knn.predict(X_test_scaled)

results_df = pd.DataFrame({'Actual': y_test, 'Predicted': y_pred})
print(results_df)

correct = (y_test.values == y_pred).sum()
incorrect = (y_test.values != y_pred).sum()
print(f'Correct: {correct}, Incorrect: {incorrect}')
```

```python
from sklearn.metrics import precision_recall_fscore_support

experiments = [
    (KNeighborsClassifier(n_neighbors=5), X_train_scaled, X_test_scaled),
    (KNeighborsClassifier(n_neighbors=5), X_train, X_test),
    (KNeighborsClassifier(n_neighbors=1), X_train_scaled, X_test_scaled),
    (KNeighborsClassifier(n_neighbors=3), X_train_scaled, X_test_scaled),
    (KNeighborsClassifier(n_neighbors=9), X_train_scaled, X_test_scaled),
]

summary = []
for i, (model, tr, te) in enumerate(experiments, 1):
  model.fit(tr, y_train)
  pred = model.predict(te)
  acc = accuracy_score(y_test, pred)
  p, r, f, _ = precision_recall_fscore_support(
      y_test, pred, average='weighted', zero_division=0
  )
  summary.append(
      {'Experiment': i, 'Accuracy': acc, 'Precision': p, 'Recall': r, 'F1': f}
  )

print(pd.DataFrame(summary))
```

```python
from sklearn.metrics import confusion_matrix

print(confusion_matrix(y_test, knn.predict(X_test_scaled)))
```
